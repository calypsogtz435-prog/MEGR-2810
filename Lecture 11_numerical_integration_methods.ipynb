## Trapezoidal Rule

### Algorithm:
1. Divide the interval [a, b] into n subintervals of equal width h = (b - a)/n.
2. Compute the sum: Integral ≈ (h/2) * [f(x₀) + 2f(x₁) + 2f(x₂) + ... + 2f(xₙ₋₁) + f(xₙ)].
3. Return the result.


```python
# Your code here
import numpy as np

# Defining function and interval
def f(x):
    return x**2

a,b = 0, 1
n = 10 
x = np.linspace(a, b, n+1) 

def trapezoidal_rule(f, a, b, n):
    h = (b - a) / n
    result = f(a) + f(b)
    for i in range(1, n):
        result += 2 * f(a + i * h)
        return result * (h / 2)


integral_approx = trapezoidal_rule(f, a, b, n)

print(f"The approximate integral is: {integral_approx}")
    
```


## Simpson's Rule

### Algorithm:
1. Divide the interval [a, b] into an even number n of subintervals.
2. Compute the sum: Integral ≈ (h/3) * [f(x₀) + 4f(x₁) + 2f(x₂) + 4f(x₃) + ... + 4f(xₙ₋₁) + f(xₙ)].
3. Return the result.


```python
# Tour code here
import numpy as np

a, b = 0, 1
n = 10
x = np.linspace(a, b, n+1)

def simpsons_rule(f, a, b, n):
    if n % 2: 
        raise ValueError("n must be even")
    h = (b - a) / n 
    y = f(x)
    sum_even = np.sum(y[2:n:2])
    sum_odd = np.sum(y[1:n:2])
    integral = (h/3) * (y[0] + 4 * sum_odd + 2 * sum_even + y[n])
    return integral

result = simpsons_rule(f, a, b, n)
print(f"The integral of x^2 from 0 to 1 is approximately: {result}")
```
